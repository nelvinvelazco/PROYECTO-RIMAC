```python
import pandas as pd
from datetime import datetime
import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")
import os
import numpy as np
pd.set_option('display.max_columns', None)
```

```python
df_declarados= pd.read_excel('C:/data/ERRORES-CONTICASA/DECLARACION-CONTICASA/Renovación de Coberturas _Conticasa_04-2025_Declarado_05-2025.xlsx', 
                             sheet_name='Hoja1', dtype={'N° Certificado Banco': str} )
```

```python
df_declarados= df_declarados.rename(columns={'N__CERTIFICADO_BANCO': 'CERTIFICADO_BANCO',
                                             'CANTIDAD_DE_COBERTURAS_A_RENOVAR': 'CANT_COBERTURAS',
                                             'P_LIZA_Y_CERTIF_': 'POLIZA_Y_CERTIFICADO'})
```
