### Conectarse a una instancia Mysql como la libreria sqlalchemy

```python
from sqlalchemy import create_engine, text

host="db-rds.czq6wkk6af5k.us-east-2.rds.amazonaws.com"
user="admin"
password=""
port=3306
database="db_pruebas"

# Conexión SQLAlchemy usando el driver de pymysql
engine = create_engine(f"mysql+pymysql://{user}:{password}@{host}:{port}/{database}")
```

```python
df_productos = pd.read_csv("C:/Data/products.csv")
df_productos.head()
```

```python
df_productos.info()
```

```python
df_productos.to_sql(
    name='productos',
    con=engine,
    if_exists='replace',  # 'replace' la reemplaza si ya existe, 'append' agrega registros
    index=False
)
```

```python
with engine.connect() as conn:
    result = conn.execute(text("SHOW TABLES;"))
    for row in result:
        print(row)
```

```python
with engine.connect() as conn:
    result = conn.execute(text("SELECT * FROM productos;"))
    for row in result:
        print(row)
```
