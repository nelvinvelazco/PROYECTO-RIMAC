```python
import pandas as pd
from datetime import datetime
import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")
import os
import numpy as np
pd.set_option('display.max_columns', None)
```

```python
df_errores1= pd.read_excel('C:/data/TABLA_ERRORES_COMPLETO.xlsx', sheet_name='Hoja1', dtype={'IDEERROR': str, 'CODERROR': str})
df_errores2 = pd.read_excel('C:/data/tabla_errores_sergio.xlsx', sheet_name='Sheet1', dtype={'IDEERROR': str, 'CODIGO_ERROR': str})
```

```python
df_errores1 = df_errores1[df_errores1.columns[:4]]
df_errores1= df_errores1.rename(columns={'CODERROR': 'CODIGO_ERROR','NOMERROR':'NOMBRE_ERROR_SAS','DSCERROR':'DESC_ERROR_SAS'})
df_errores1['NOMBRE_ERROR_SAS']= df_errores1['NOMBRE_ERROR_SAS'].str.upper()
df_errores1['DESC_ERROR_SAS']= df_errores1['DESC_ERROR_SAS'].str.upper()
df_errores1['IDEERROR'] = df_errores1['IDEERROR'].apply(lambda x: str(int(x)).zfill(4) if pd.notnull(x) else "")
df_errores1['CODIGO_ERROR'] = df_errores1['CODIGO_ERROR'].apply(lambda x: str(int(x)).zfill(4) if pd.notnull(x) else "")
```

```python
df_errores1= df_errores1.merge(df_errores2, on='IDEERROR', how='left')
```

```python
df_errores1= df_errores1.rename(columns={'CODIGO_ERROR_x': 'CODIGO_ERROR'})
df_errores1.drop(columns=['CODIGO_ERROR_y'], inplace=True)
```
