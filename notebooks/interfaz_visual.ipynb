```python
import ipywidgets as widgets
from IPython.display import display

fecha_inicio = widgets.DatePicker(description='Fecha inicio')
fecha_fin = widgets.DatePicker(description='Fecha fin')

boton = widgets.Button(description="Ejecutar")

def on_button_click(b):
    fi = fecha_inicio.value
    ff = fecha_fin.value
    
    print("Fecha inicio:", fi)
    print("Fecha fin:", ff)

boton.on_click(on_button_click)

display(fecha_inicio, fecha_fin, boton)
```

```python
import ipywidgets as widgets
from IPython.display import display

selector = widgets.SelectMultiple(
    options=['OPCION_1', 'OPCION_2', 'OPCION_3'],
    description='Opciones'
)

boton = widgets.Button(description="Procesar")

def on_click(b):
    seleccionados = list(selector.value)
    
    print("Seleccionados:", seleccionados)
    
    for item in seleccionados:
        print("Procesando:", item)

boton.on_click(on_click)

display(selector, boton)
```

```python
import ipywidgets as widgets
from IPython.display import display

lista1 = ['DAR', 'FCR1', 'FCR2','OMX','PIC']
lista2 = ['RECLAMOS', 'TICKET INFORMACION', 'REPORTE EMISION SATA','LA']
label = widgets.Label("Seleccione opciones:")
checkboxes1 = [
    widgets.Checkbox(value=True, description=op)
    for op in lista1
]

checkboxes2 = [
    widgets.Checkbox(value=True, description=op)
    for op in lista2
]

contenedor = widgets.GridBox(
    checkboxes1 + checkboxes2,
    layout=widgets.Layout(grid_template_columns="repeat(5, 180px)")
)
boton = widgets.Button(
    description="Ejecutar proceso",
    button_style='success'
)
display(label, contenedor,boton)
```

```python
import ipywidgets as widgets
from IPython.display import display

label1 = widgets.Label("Seleccione fechas:")
label2 = widgets.Label("Seleccione bases:")
# 🔹 Fechas
fecha_inicio = widgets.DatePicker(description='Inicio')
fecha_fin = widgets.DatePicker(description='Fin')

# 🔹 Opciones (todas seleccionadas por defecto)
opciones = ['REPORTE_1', 'REPORTE_2', 'REPORTE_3']

checkboxes = [
    widgets.Checkbox(value=True, description=op)
    for op in opciones
]

contenedor = widgets.HBox(
    checkboxes,
    layout=widgets.Layout(flex_flow='row wrap')
)

# 🔹 Botón
boton = widgets.Button(
    description="Ejecutar proceso",
    button_style='success'
)

# 🔹 Lógica principal (AQUÍ VA TU SCRIPT)
def ejecutar(b):
    fi = fecha_inicio.value
    ff = fecha_fin.value
    seleccionados = [cb.description for cb in checkboxes if cb.value]

    # Validaciones
    if fi is None or ff is None:
        print("❌ Debes seleccionar ambas fechas")
        return
    
    if not seleccionados:
        print("❌ Debes seleccionar al menos una opción")
        return

    print("📅 Fechas:", fi, "→", ff)
    print("📌 Procesos seleccionados:", seleccionados)

    # 🔥 Aquí recorres tu lista
    for proceso in seleccionados:
        print(f"🚀 Ejecutando {proceso}...")
        
        # 👉 Aquí puedes llamar tus funciones reales
        # ejemplo:
        # if proceso == 'REPORTE_1':
        #     procesar_reporte_1(fi, ff)
        # elif proceso == 'REPORTE_2':
        #     procesar_reporte_2(fi, ff)

boton.on_click(ejecutar)

# 🔹 UI final
panel = widgets.VBox([
    #widgets.HTML("<h3>Parámetros del Proceso</h3>"),
    label1,
    fecha_inicio,
    fecha_fin,
    label2,
    contenedor,
    boton
])

display(panel)
```
